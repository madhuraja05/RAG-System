```python
from pathlib import Path

folder_path = Path("/content/drive/MyDrive/RAG system")

print("Folder exists:", folder_path.exists())
```

```python
pdf_files = list(folder_path.glob("*.pdf"))

print("PDF files found:")

for pdf in pdf_files:
    print(pdf.name)
```

```python
from pypdf import PdfReader

pdf_file = pdf_files[0]

reader = PdfReader(pdf_file)

print("PDF:", pdf_file.name)
print("Number of pages:", len(reader.pages))
```

```python
all_text = ""

for page in reader.pages:
    text = page.extract_text()
    if text:
        all_text += text + "\n"

print("Text extracted successfully!")
print(all_text[:2000])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

chunks = text_splitter.create_documents([all_text])

print("Total chunks:", len(chunks))
print("\nFirst chunk:")
print(chunks[0].page_content)
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded successfully!")
```

```python
texts = [chunk.page_content for chunk in chunks]

embeddings = embedding_model.encode(
    texts,
    show_progress_bar=True
)

print("Number of chunks:", len(texts))
print("Number of embeddings:", len(embeddings))
print("Embedding size:", len(embeddings[0]))
```

```python
import chromadb

chroma_client = chromadb.Client()

collection = chroma_client.get_or_create_collection(
    name="rag_documents"
)

print("Vector database created successfully!")
```

```python
import uuid

ids = [str(uuid.uuid4()) for _ in texts]

collection.add(
    ids=ids,
    documents=texts,
    embeddings=embeddings.tolist()
)

print("Chunks stored in Vector DB:", collection.count())
```

```python
question = "What is a Common Table Expression in SQL?"

print("Question:", question)
```

```python
question_embedding = embedding_model.encode(
    [question]
)[0]

print("Question embedding created!")
print("Embedding size:", len(question_embedding))
```

```python
results = collection.query(
    query_embeddings=[question_embedding.tolist()],
    n_results=3
)

print("Relevant chunks retrieved:", len(results["documents"][0]))
```

```python
retrieved_chunks = results["documents"][0]

for i, chunk in enumerate(retrieved_chunks, 1):
    print(f"\n--- Retrieved Chunk {i} ---")
    print(chunk)
```

```python
context = "\n\n".join(retrieved_chunks)

print("Context created successfully!")
print(context[:3000])
```

```python
from google.colab import userdata
from groq import Groq

GROQ_API_KEY = userdata.get("GROQ_API_KEY")

groq_client = Groq(api_key=GROQ_API_KEY)

print("Groq connected successfully!")
```

```python
prompt = f"""
You are a helpful SQL assistant.

Answer the question using only the information provided in the context.

If the answer is not present in the context, say:
"I cannot find the answer in the document."

Context:
{context}

Question:
{question}

Answer:
"""

response = groq_client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {
            "role": "user",
            "content": prompt
        }
    ],
    temperature=0
)

answer = response.choices[0].message.content

print("Final Answer:")
print(answer)
```

```python
def ask_rag(question):
    question_embedding = embedding_model.encode([question])[0]

    results = collection.query(
        query_embeddings=[question_embedding.tolist()],
        n_results=3
    )

    retrieved_chunks = results["documents"][0]

    context = "\n\n".join(retrieved_chunks)

    prompt = f"""
You are a helpful SQL assistant.

Answer the question using only the information provided in the context.

If the answer is not present in the context, say:
"I cannot find the answer in the document."

Context:
{context}

Question:
{question}

Answer:
"""

    response = groq_client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0
    )

    return response.choices[0].message.content
```

```python
question = "What is a Common Table Expression in SQL?"

answer = ask_rag(question)

print("Question:", question)
print("\nAnswer:", answer)
```
